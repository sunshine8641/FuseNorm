```python
import numpy as np
from scipy.stats import gumbel_r
from scipy.special import expit  # sigmoid


# plot_histograms([-np.std(scores_in_all_i_fuse, axis=0)], [-np.std(scores_out_all_i_fuse, axis=0)], bins=50, title="Histograms of OOD Scores")

def compute_norm_shift(scores_test, means, left_q=0, right_q=0,mode='0'):
    """
    :param scores_test: shape (num_layers, N)
    :param means: (num_layers,)
    :param stds_left: (num_layers,)  # 左侧std (小于mean部分)
    :param stds_right: (num_layers,) # 右侧std (大于mean部分)
    """

    z_scores = np.zeros_like(scores_test)
    if mode=='0':
        z_scores =np.abs((scores_test - means[:, None]) / stds[:, None])  # shape (num_layers, N)
        return z_scores
    elif mode=='1':
        for i in range(len(scores_test)):
            mean = means[i]
            left = left_q[i]
            right = right_q[i]
            x = scores_test[i]
            z = np.zeros_like(x)

            # 大于 mean 的用右侧 std
            mask_right = x >= mean
            z[mask_right] = (x[mask_right] - mean) / right

            # 小于 mean 的用左侧 std
            mask_left = x < mean
            z[mask_left] = (mean - x[mask_left]) / left

            z_scores[i] = z
        return z_scores  # shape (num_layers, N)
    if mode=='2':
        z_scores =(scores_test - means[:, None]) / stds[:, None]  # shape (num_layers, N)
        return z_scores

def compute_ood_score(scores_test,mode='0'):
    if mode=='0':
        return -np.max(scores_test, axis=0)

    elif  mode=='1':
        return -np.std(scores_test, axis=0)


    # final_scores = -np.max(z_scores, axis=0)  # 最偏离的一层决定

    # return final_scores
    
# def compute_ood_score(scores_test, means, stds, mode="max"):
#     """
#     基于z-score计算OOD分数
#     :param scores_test: shape (num_layers, N)，待测样本的分数
#     :param means, stds: 来自ID validation set 的统计
#     :param mode: 融合方式，可选 ["max", "sum", "avg"]
#     :return: OOD scores (N,)
#     """
#     # 计算逐层 z-score 偏离度
#     z_scores = np.abs((scores_test - means[:, None]) / stds[:, None])  # shape (num_layers, N)

#     if mode == "max":
#         # final_scores = expit(-np.max(z_scores, axis=0)+3)   # 最偏离的一层决定
#         final_scores = -np.max(z_scores, axis=0)  # 最偏离的一层决定

#     elif mode == "sum":
#         final_scores = -np.sum(z_scores, axis=0)   # 累积异常
#     elif mode == "avg":
#         final_scores = -np.mean(z_scores, axis=0)  # 平均异常
#     elif mode=="gumbel":
#         return compute_ood_score_gumbel(scores_test, means, stds)
#     else:
#         raise ValueError("mode must be one of ['max', 'sum', 'avg']")

#     return final_scores
```

```python
def unnormalize(t,mean,std):
    mean = torch.tensor(mean).view(3, 1, 1)
    std = torch.tensor(std).view(3, 1, 1)
    return torch.clamp(t * std + mean, 0, 1)

def show_images(x, y,mean,std, n=16,save_path=None):
    x = x[:n]
    y = y[:n]
    x = torch.stack([unnormalize(img.cpu(),mean,std) for img in x])
    grid_img = transforms.ToPILImage()(T.make_grid(x, nrow=int(n**0.5)))
    plt.figure(figsize=(4,4))
    plt.imshow(grid_img)
    plt.axis("off")
    plt.title("Sample Images")
    plt.savefig(save_path, dpi=100, bbox_inches="tight")
    plt.show()
```

```python
def compute_id_statistics(scores_in):
    """
    scores_in:  L*N,  [[s_1,s2,..sN] ,...]
    """
    
    num_layers = len(scores_in)
    stds = np.std(scores_in, axis=1) + 1e-8  # 每层标准差，防止除零

    means = np.means(scores_in,axis=1)

    return means, stds
```
